```python
import requests
import pandas as pd

url = "https://restcountries.com/v3.1/all"

try:
    # Task 1 — Fetch and Validate
    response = requests.get(url)
    print("Status Code:", response.status_code)

    if response.status_code == 200:
        data = response.json()
        print("Total countries retrieved:", len(data))

        # Task 2 — Extract Fields
        country_list = []

        for country in data:
            name = country.get("name", {}).get("common", "N/A")
            population = country.get("population", 0)
            region = country.get("region", "N/A")
            area = country.get("area", 0)

            country_list.append({
                "common_name": name,
                "population": population,
                "region": region,
                "area": area
            })

        df = pd.DataFrame(country_list)

        print("\nFirst 10 rows:")
        print(df.head(10))

        # Task 3 — Summary
        region_population = df.groupby("region")["population"].sum()
        top_region = region_population.idxmax()

        largest_country = df.loc[df["area"].idxmax(), "common_name"]

        print("\nRegion with highest total population:", top_region)
        print("Country with largest area:", largest_country)

    else:
        print("Failed to fetch data. Please check the API.")

except Exception as e:
    print("Error occurred:", e)
```
